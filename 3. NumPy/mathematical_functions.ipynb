```python
import numpy as np
```

```python
#Mathematical fnx

#Aggregation
arr = np.array([1, 2, 3, 4, 5, 6])
print(np.sum(arr))
print(np.prod(arr))
print(np.min(arr))
print(np.max(arr))
print(np.argmin(arr)) #index of min element
print(np.argmax(arr)) #index of max element

print(np.mean(arr))
print(np.std(arr))
print(np.median(arr))
print(np.var(arr))
```

```python
#Power fnx

arr = np.array([1, 2, 3, 4, 5, 6])

print(np.square(arr))
print(np.sqrt(arr))
print(np.pow(arr, 3))
```

```python
#log

arr = np.array([1, 2, 3, 4, 5, 6])
print(np.log(arr))
print(np.log2(arr))
print(np.log10(arr))
print(np.exp(arr))
```

```python
#Rounding

print(np.round(3.14))
print(np.ceil(3.14))
print(np.floor(3.14))
print(np.trunc(3.14))

arr = np.array([1, -1, 3, 3, -9, 6])

print(np.unique(arr))
print(np.sort(arr))

print(np.abs(arr))
```
