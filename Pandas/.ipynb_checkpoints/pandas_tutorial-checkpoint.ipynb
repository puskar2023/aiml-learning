```python
import pandas as pd
```

```python
info = {
    "Name": ["Sarthak", "Harsh", "Ankur"],
    "CGPA": [8.5, 7.5, 9.5]
}
df = pd.DataFrame(info)
print(df)
```

```python
s = pd.Series([1, 2, 3, 4, 5, 6])
print(s)
print(type(s))

# index
print(s[0])
print(s[2])
```

```python
s = pd.Series([23, 24, 25, 26], index=["Adam", "Bob", "Eve", "Charlie"])
print(s)
print(s["Adam"])
print(s.index)
```

```python
s1 = pd.Series([1, 2, 3, 4, 5, 6])
s2 = pd.Series([10, 11, 12, 13, 14, 15])

print(s1 + s2)

s1[0] = 100
changed_s1 = s1.drop(0)
print(s1)
print(changed_s1)
```

```python
#Types of creating DataFrame

#Type1

info = {
    "Name": ["Adam", "Bob", "Eve"],
    "Age": [23, 26, 29],
    "GPA": [9.1, 9.2, 9.3]
}

df = pd.DataFrame(info)
print(df)
print(df.index)
print(df.columns)
```

```python
#Type2

df = pd.DataFrame([["Adam", 23], ["Bob", 24], ["Eve", 29]], columns = ["Name", "Age"])
print(df)
```

```python
#Type 3

import numpy as np

np_arr = np.array([[1, 2, 3], [4, 5, 6]])
df = pd.DataFrame(np_arr, columns= ["A", "B", "C"] )
print(df)
```

```python
#AQI Data Set

df = pd.read_csv("globalAirQuality.csv")

df.describe()
```

```python
# selecting columns

#columns
df["city"]
df[["city", "aqi"]]

#rows
df.loc[2]
df.loc[0:2] #start index : end index (inclusive)

df.iloc[2]
df.iloc[0:2] #start index : end index (exclusive)

#cells - row, columns
df.loc[0:2, "city"]
df.columns

df.loc[0:2, ["country", "city", "latitude", "longitude"]]
df.iloc[0:3, 1:5]

#cell
df.at[0, "city"]
df.iat[0, 2]

#copy
cities = df["city"].copy()
cities[0] = "Hello"
```
