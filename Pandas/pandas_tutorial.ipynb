```python
#Cleaning data - missing value
df = pd.read_csv("raw_data.csv")

df.isnull()
df.isna()

df.isnull().sum()

df.dropna() #delete missing value rows
df.dropna(axis = 1) #delete missing value columns

df.fillna(0)  #fill missing values with 0

age_mean = df["age"].mean()

cleaned_df = df.copy()
cleaned_df["age"] = cleaned_df["age"].fillna(age_mean)


df.ffill()  #forward fill -> replace null values with previous values
df.bfill()  #backward fill -> replace null values with next values
```

```python
#Handle Duplicates

df.duplicated()
df2 = df.copy()
df2.drop_duplicates(inplace=True)
```

```python
#Data Types
df.dtypes

df2 = df.copy()
df2 = df2.fillna(0)
df2["age"] = df2["age"].astype("int64")
df2.dtypes

df2["gender"].str.lower()
df2["gender"].str.upper()
df2["gender"].str.capitalize()

df2["name"].str.split(" ")
df2["country"].str.contains("india", case=False)
```

```python
df2 = df.copy()
df2["tax"] = df["income"].apply(lambda x: "20%" if x >= 50000 else "10%")

gender_map = {"Male": "M", "Female": "F"}
df2["gender"] = df2["gender"].map(gender_map)

df2 = df2.assign(new_income = df2["income"]*1.1)

df2["country"] = df2["country"].replace("USA", "US")

df2.columns = df2.columns.str.capitalize()
df2.rename(columns={"Income": "Salary"})
df2.rename(index={1: "First"})

# map()      → map values according to a mapping
# replace()  → replace particular values
# apply()    → apply a function/logic
# assign()   → add/create a column
# rename()   → rename columns/index

df2 = df2.fillna(50)
# sorted_df = df2.sort_values("Income", ascending=False)
sorted_df = df2.sort_values(["Income", "Age"])

sorted_df.sort_index()

# reset
sorted_df.reset_index()
sorted_df.reset_index(drop=True)

# Ranking
sorted_df["Ranking"] = sorted_df["Income"].rank(ascending=False, method="dense") #method="min/max"
sorted_df
```
